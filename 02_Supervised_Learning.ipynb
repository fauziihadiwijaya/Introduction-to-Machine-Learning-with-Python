## What is Supervised Learning?

Supervised learning is a machine learning method that uses labeled data.

This means the dataset already has answers (target labels), so the model learns from input and output pairs.

### Examples:
- Predict house prices
- Detect spam emails
- Predict pass/fail students

## Types of Supervised Learning

### 1. Classification
Predict categories.

Examples:
- Spam / Not Spam
- Sick / Healthy

### 2. Regression
Predict numeric values.

Examples:
- House price
- Sales prediction

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
```

## Example: Classification using K-Nearest Neighbors (KNN)
We use Breast Cancer dataset from sklearn.

```python
# Load dataset
cancer = load_breast_cancer()

# Convert to DataFrame
df = pd.DataFrame(cancer.data, columns=cancer.feature_names)

df.head()
```

```python
# Split data
X_train, X_test, y_train, y_test = train_test_split(
    cancer.data,
    cancer.target,
    random_state=42
)
```

```python
# Build model
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
```

```python
# Accuracy score
print("Training Accuracy:", knn.score(X_train, y_train))
print("Testing Accuracy :", knn.score(X_test, y_test))
```

```python
scores = pd.DataFrame({
    "Type": ["Training", "Testing"],
    "Score": [
        knn.score(X_train, y_train),
        knn.score(X_test, y_test)
    ]
})

plt.figure(figsize=(8,4))

sns.barplot(
    data=scores,
    x="Type",
    y="Score",
    palette=["#FF7675","#74B9FF"]
)

plt.ylim(0.85,1.0)
plt.title("KNN Accuracy Comparison", fontsize=14, weight="bold")

plt.show()
```

### Interpretation

Training and testing accuracy are both high and close in value.

This indicates that the KNN model generalizes well and does not strongly overfit.

```python
labels = ["Correct Prediction", "Error"]
values = [93,7]

colors = ["#00CEC9","#E17055"]

plt.figure(figsize=(6,6))

plt.pie(values,
        labels=labels,
        autopct="%1.0f%%",
        colors=colors,
        startangle=90)

plt.title("Prediction Performance")
plt.show()
```

### Insight

Most predictions are correct, showing that the model performs reliably on this dataset.

## Interpretation
