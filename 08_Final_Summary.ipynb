```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

## Model Accuracy Comparison

```python
models = pd.DataFrame({
    "Model": [
        "KNN",
        "Random Forest",
        "Naive Bayes",
        "Pipeline KNN",
        "Logistic Regression"
    ],

    "Accuracy": [
        0.93,
        0.96,
        0.91,
        0.95,
        0.94
    ]
})

models
```

```python
plt.figure(figsize=(10,5))

colors = ["#FF6B6B","#FFD93D","#6BCB77","#4D96FF","#9D4EDD"]

bars = plt.barh(models["Model"], models["Accuracy"], color=colors)

plt.xlim(0.85,1.0)
plt.title("Model Accuracy Comparison", fontsize=15, weight="bold")
plt.xlabel("Accuracy")

for bar in bars:
    plt.text(bar.get_width()+0.002,
             bar.get_y()+0.25,
             round(bar.get_width(),2),
             fontsize=10)

plt.show()
```

```python
labels = models["Model"]
stats = models["Accuracy"]

angles = np.linspace(0, 2*np.pi, len(labels), endpoint=False)
stats = np.concatenate((stats,[stats[0]]))
angles = np.concatenate((angles,[angles[0]]))

fig = plt.figure(figsize=(7,7))
ax = fig.add_subplot(111, polar=True)

ax.plot(angles, stats, linewidth=2)
ax.fill(angles, stats, alpha=0.25)

ax.set_thetagrids(angles[:-1] * 180/np.pi, labels)
ax.set_title("Radar View of Model Performance", pad=20)

plt.show()
```

```python
topics = ["Supervised","Unsupervised","Evaluation","Pipeline","Text Data"]
values = [30,20,20,15,15]

colors = ["#ff9f1c","#2ec4b6","#e71d36","#8338ec","#3a86ff"]

plt.figure(figsize=(7,7))

plt.pie(
    values,
    labels=topics,
    autopct="%1.0f%%",
    startangle=90,
    colors=colors,
    wedgeprops={'width':0.4}
)

plt.title("Learning Focus Distribution")
plt.show()
```

```python
chapters = ["Ch2","Ch4","Ch5","Ch6","Ch7"]
scores = [0.93,0.93,0.95,0.96,0.91]

plt.figure(figsize=(10,4))

plt.plot(chapters, scores, marker="o", linewidth=3)

plt.title("Performance Across Chapters")
plt.ylabel("Accuracy")
plt.ylim(0.88,1.0)

for i,v in enumerate(scores):
    plt.text(i, v+0.003, str(v))

plt.show()
```
