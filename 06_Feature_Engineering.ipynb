## What is Feature Engineering?

Feature engineering is the process of creating, selecting, or transforming variables (features) to improve machine learning model performance.

Good features often produce better predictions than complex models.

## Common Techniques

### 1. Feature Selection
Choose the most important variables.

### 2. Feature Scaling
Normalize feature values.

### 3. Feature Creation
Create new variables from existing data.

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer
from sklearn.ensemble import RandomForestClassifier
```

```python
# Load dataset
cancer = load_breast_cancer()

X = pd.DataFrame(cancer.data, columns=cancer.feature_names)
y = cancer.target

X.head()
```

## Train Random Forest Model

We use Random Forest to identify important features.

```python
model = RandomForestClassifier(random_state=42)
model.fit(X, y)
```

```python
importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

importance = importance.sort_values(by="Importance", ascending=False)
importance.head(10)
```

## Visual Feature Importance

```python
plt.figure(figsize=(10,6))

sns.barplot(
    data=importance.head(10),
    x="Importance",
    y="Feature"
)

plt.title("Top 10 Important Features")
plt.show()
```

## Interpretation

Some features contribute more strongly to prediction results.

By selecting important features, we can simplify models and improve efficiency.

## Conclusion

- Feature engineering improves model quality.
- Important features help reduce noise.
- Random Forest can estimate feature importance.
