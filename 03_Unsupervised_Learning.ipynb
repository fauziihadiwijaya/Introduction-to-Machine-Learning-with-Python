## What is Unsupervised Learning?

Unsupervised learning is a machine learning method that uses unlabeled data.

The model tries to find hidden patterns or groupings in the data automatically.

## Common Types

### 1. Clustering
Grouping similar data points.

Example:
- Customer segmentation

### 2. Dimensionality Reduction
Reducing number of features while keeping important information.

Example:
- PCA

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
```

## Example Dataset

We create synthetic data with several groups.

```python
# Create dataset
X, y = make_blobs(
    n_samples=300,
    centers=3,
    random_state=42
)

plt.scatter(X[:,0], X[:,1])
plt.title("Original Data")
plt.show()
```

## Apply K-Means Clustering

```python
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)

labels = kmeans.labels_
```

```python
plt.figure(figsize=(9,6))

sns.scatterplot(
    x=X[:,0],
    y=X[:,1],
    hue=labels,
    palette="Set2",
    s=100
)

plt.title("K-Means Clustering Result", fontsize=14, weight="bold")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")

plt.legend(title="Cluster")
plt.show()
```

### Interpretation

The data points are separated into three distinct clusters.

This means K-Means successfully identified natural groupings in the dataset.

```python
cluster_df = pd.DataFrame({"Cluster": labels})

plt.figure(figsize=(7,4))

sns.countplot(
    data=cluster_df,
    x="Cluster",
    palette="Set2"
)

plt.title("Number of Data Points per Cluster")
plt.show()
```

### Insight

Each cluster contains a relatively balanced number of observations, indicating stable segmentation.

## Interpretation

K-Means groups data into clusters based on similarity.

Each color represents one cluster.

## Conclusion

- Unsupervised learning uses unlabeled data.
- Clustering groups similar data.
- K-Means is one of the most common clustering algorithms.
